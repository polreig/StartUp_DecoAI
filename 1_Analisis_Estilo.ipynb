Importar librerías y guardar API key

```python
from google import genai
from google.colab import userdata
from PIL import Image
import requests
from io import BytesIO

# Cargamos la clave de forma segura
GOOGLE_API_KEY = userdata.get('clave_API_gemini')
client = genai.Client(api_key=GOOGLE_API_KEY)
```

Función para analizar la habitación

```python
def analizar_habitacion(ruta_imagen_o_url, texto_usuario):
    # Cargar la imagen
    if ruta_imagen_o_url.startswith('http'):
        response = requests.get(ruta_imagen_o_url)
        imagen = Image.open(BytesIO(response.content))
    else:
        imagen = Image.open(ruta_imagen_o_url)

    # Creamos el Prompt
    prompt = f"""
    Eres un diseñador de interiores profesional.
    Analiza la imagen adjunta de esta habitación actual y el siguiente mensaje del cliente: '{texto_usuario}'.

    Por favor, responde con la siguiente estructura:
    1. Estilo Actual: Describe en una frase el estilo y estado actual de la habitación.
    2. Recomendación de Estilo: Basado en lo que pide el cliente, ¿qué estilo específico le recomiendas?
    3. Elementos Clave: Enumera 3 muebles o decoraciones esenciales para lograr la transformación.
    4. Paleta de Colores: Sugiere 3 colores para el nuevo estilo.
    """

    # 3. NUEVA forma de llamar al modelo
    respuesta = client.models.generate_content(
        model='gemini-2.5-flash',
        contents=[imagen, prompt]
    )

    return respuesta.text
```

Prueba

```python
# Cambia esta URL por la de una foto que os guste para probar
url_foto_prueba = "https://concept-u.es/cdn/shop/articles/habitacion-comoda.webp?v=1703692446"
peticion_cliente = "Quiero que la habitación sea más colorida"

resultado = analizar_habitacion(url_foto_prueba, peticion_cliente)
print(resultado)
```
